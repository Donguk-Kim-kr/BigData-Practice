# 2) 데이터 병합

```python
city = happy_life["시도"].unique()
city
```

```python
# 리스트로 변환
city = list(city)
city
```

```python
# 데이터프레임으로 생성
happy_merge = pd.DataFrame({
    "시도": city
})
happy_merge.head()
```

```python
happy_life.head()
```

## 시도별 삶의 만족도 평균

```python
# 시도별 삶의 만족도 평균
life = happy_life["삶의 만족도"].groupby(by=happy_life["시도"]).mean()
life    # 시리즈
```

```python
# 시도를 중심으로 병합한다.
happy_merge = pd.merge(happy_merge, life, on="시도")
happy_merge.head()
```

```python
happy_health.head()
```

## 시도별 건강 평균

```python
# 시도별 건강 평균
health = happy_health["평균"].groupby(happy_health["시도"]).mean()
```

```python
health
```

```python
happy_merge = pd.merge(happy_merge, health, on="시도")
happy_merge.head()
```

```python
happy_merge = pd.merge(happy_merge, health.rename("건강"), on="시도")
happy_merge.head()
```

```python
happy_merge = happy_merge.drop("평균", axis=1)
```

```python
happy_merge.head()
```

# 4) 시각화

```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns

# 한글 폰트 설정
font_path = ("C:\Windows\Fonts\KyoboHandwriting2020pdy.ttf")
fm.fontManager.addfont(font_path)
font_name = fm.FontProperties(fname=font_path, size=16).get_name()

plt.rc("font", family=font_name)
```

```python
happy_merge.columns
```

```python
plt.figure(figsize=(15, 6)) # 가로 15인치, 세로 6인치 비율
items = list(happy_merge.columns[1:9])
for a in items:
    chartdata = happy_merge[a]
    plt.plot(chartdata, marker="o", label=a)

plt.xlabel("시도") # x축 제목
plt.ylabel("수치") # y축 제목
plt.xticks(range(17), happy_merge["시도"], fontsize=8)
plt.title("대한민국 행복지수") # 차트 제목
plt.legend() # 범례
plt.grid()
plt.show()
```
