## 불(bool) 인덱싱

```python
w = df['temp'] >= 30 # 부등호를 먼저 체크하고 변수에 담긴다.
df[w] # 참(True)인 것만 나온다.
```

```python
w
```

```python
df[df['temp'] >= 30]
```

### 조건에 맞는 데이터 추출

```python
# 가장 더웠던 날의 모든 정보(행)
w = df['temp'] == df['temp'].max()
df[w]
```

```python
df[df['temp'] == df['temp'].max()]
```

### 조건이 2개 이상 --> 그리고 &, 또는 |

```python
# 기온이 30도 이상이고, 최대풍속이 9m/s이상인 데이터

w = (df['temp'] >= 30) & (df['max_wind'] >= 9)
df[w]
```

```python
# 기온이 30도 이상이거나 최대풍속이 9m/s이상인 데이터

w = (df['temp'] >= 30) | (df['max_wind'] >= 9)
df[w]
```

## isna() --> 값이 nan(결측치)이라면 True

```python
# 최대 풍속이 측정되지 않은 모든 데이터 출력
df['max_wind'].isna()
```

## 데이터프레임명.isnull().sum() --> 컬럼별 결측데이터 개수 확인(합으로 확인)

- 0 False 거짓
- 1 True 참

```python
df.isnull()
```

```python
df.isnull().sum() # 컬럼별 결측데이터 수
```

```python
df.isnull().sum().sum() # 전체 데이터프레임 결측데이터 수
```

## 결측데이터 대체하기 --> fillna(채울 값)

```python
df['max_wind'].isnull().sum()
```

```python
df3 = df.copy() # 복사
df3
```

```python
df3 = df3.fillna(0)
```

```python
df3.isnull().sum()
```

```python
# 평균 --> mean()
avg = df3['max_wind'].mean()
avg
```

```python
df3['max_wind'] = df3['max_wind'].fillna(avg)
```

```python
avg2 = df3['mean_wind'].mean()
avg2
```

```python
df3['mean_wind'] = df3['mean_wind'].fillna(avg2)
```
