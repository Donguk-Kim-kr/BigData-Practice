# 리스트의 for문

```python
players = ["채치수", "권준호", "이달재", "송태섭", "신오일", "강백호", "서태웅", "정대만"]
```

```python
for p in players:
    print(p) # 리스트의 값이 나온다.
```

```python
for i in range(5):
    print(i)
```

## enumerate 함수

```python
# for 인덱스 번호, 값 in enumearate(컬렉션 자료형):
#     pass
```

```python
for p in enumerate(players):
    print(p)
```
