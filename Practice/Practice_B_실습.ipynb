```python
with open("guest.txt", "w" , encoding="utf-8") as f:
    count = 0
    while True:
        name = input("졸업 전시회 참석 학생 이름을 입력하세요. : ")
        
        if not name:
            break
        f.write(name + "\n")
        count += 1
        print(f"{name} welecome!!")
    print(f"total count : {count}")
```
