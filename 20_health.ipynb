# 2020년 건강검진 일부 데이터 --> 산점도 그래프

```python
import pandas as pd
import matplotlib.pyplot as plt
import koreanize_matplotlib
```

```python
# 데이터 불러오기

data = pd.read_excel('health_screenings_2020_1000ea.xlsx')
```

```python
data.shape
```

```python
data.head()
```

## 키, 몸무게 데이터로 산점도 그래프 그리기

```python
heightdata = data['height'] # 키
weightdata = data['weight'] # 몸무게

plt.figure(figsize=(10, 4))
plt.scatter(x=heightdata, y=weightdata)

plt.title('2020년 건강검진 산점도 그래프')
plt.xlabel('키')
plt.ylabel('몸무게')

plt.grid()
plt.show()
```

## 수축기 혈압, 이완기 혈압, 공복 혈당 그룹 산점도 그래프

```python
systolic_data = data['systolic'] # 수축기 혈압
diastolic_data = data['diastolic'] # 이완기 혈압
blood_sugar_data = data['blood_sugar'] # 공복 혈당

plt.figure(figsize=(10, 6))

plt.scatter(x=systolic_data, y=diastolic_data, color='r', edgecolors='black', label='수축기 * 이완기')
plt.scatter(x=systolic_data, y=blood_sugar_data, color='g', edgecolors='black', label='수축기 * 공복 혈당')
plt.scatter(x=diastolic_data, y=blood_sugar_data, color='b', edgecolors='black', label='이완기 * 공복 혈당')

plt.title('2020년 건강검진 그룹 산점도 그래프')
plt.legend() # 범례
plt.grid()
plt.xlim(40, 180)
plt.ylim(30, 300)

plt.show()
```

## 여성 키 히스토그램

```python
data.head()
```

```python
woman = data['gender'] == 2
woman
```

```python
female_data = data.loc[woman]
female_data.head()
```

```python
female_data = data.loc[woman, ['gender', 'height']]
female_data.head()
```

```python
plt.figure(figsize=(10, 6))

# 키의 분포를 알고 싶다. 구간을 7개로 나누었다.

plt.hist(female_data['height'], bins=7, label='Female')
plt.title('2020년 건강검진 여성 키 히스토그램')
plt.xlim(130, 180)
plt.xlabel('키')
plt.ylabel('빈도수(frequency)')
plt.legend()
plt.grid()
plt.show()
```

## 남성, 여성 키 그룹 히스토그램

```python
man = data['gender'] == 1
male_data = data.loc[man]

male_data = data.loc[man, ['gender', 'height']]

plt.figure(figsize=(10, 6))

# 키의 분포를 알고 싶다. 구간을 7개로 나누었다.

plt.hist(male_data['height'], bins=9, label='male')
plt.title('2020년 건강검진 여성 키 히스토그램')
plt.xlim(140, 200)
plt.xlabel('키')
plt.ylabel('빈도수(frequency)')
plt.legend()
plt.grid()
plt.show()
```
